```python
import polars as pl
pl.Config.set_tbl_cols(200).set_fmt_str_lengths(150).set_tbl_rows(100)
```

```python
DIR_PATH = "../data/raw_data/exp_viv_turistica_tabla5_MAY2025.csv" # "../data/raw_data/secciones_censales_pais_nacimiento.csv"
tt = pl.read_csv(DIR_PATH, 
                 separator=";",
                 schema_overrides={
                     "CUSEC": pl.Utf8}).with_columns(
                     pl.col("Porcentaje vivienda turistica").str.replace_all(",", ".").cast(pl.Float16),
                     pl.col("plazas por vivienda turistica").str.replace_all(",", ".").fill_null(0).cast(pl.Float16)
                 ).select(["CUSEC", "vivienda turistica","plazas","plazas por vivienda turistica", 
                           "Porcentaje vivienda turistica", "PERIODO", "MUN_LITERAL"])
tt.write_csv("../data/clean_data/vivienda_turistica_secciones.csv")
```

```python
renta_dir_path = "../data/raw_data/renta_ine_españa.csv"
renta = pl.read_csv(renta_dir_path,
                    separator=";").filter(
                        pl.col("Secciones").is_not_null()
                    ).with_columns(
                        pl.col("Secciones").str.extract(r"\d+{10}", 0),
                        pl.col("Total").cast(pl.Float64, strict=False).fill_null(0.0)
                    )
renta.write_csv("../data/clean_data/renta_ine_españa_limpia.csv")
```
