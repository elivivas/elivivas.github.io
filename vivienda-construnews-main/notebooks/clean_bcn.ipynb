```python
import polars as pl
pl.Config.set_tbl_cols(300).set_fmt_str_lengths(400).set_tbl_rows(20)
```

```python
code_houseing = ["45211200", "45211000", "45211100", "45211300", "45211340"]

raw = pl.read_csv(
    "../data/raw_data/Contractació_pública_a_Catalunya__publicacions_a_la_Plataforma_de_serveis_de_contractació_pública_20260317.csv"
    ).filter(
        pl.col("codi_cpv").str.starts_with(code_houseing[0]) |
        pl.col("codi_cpv").str.starts_with(code_houseing[1]) |
        pl.col("codi_cpv").str.starts_with(code_houseing[2]) |
        pl.col("codi_cpv").str.starts_with(code_houseing[3]) |
        pl.col("codi_cpv").str.starts_with(code_houseing[4]) 
    ).filter(
        pl.col("resultat").is_not_null()
    ).select(
        pl.col("codi_expedient").alias("id"),
        pl.col("objecte_contracte").alias("title"),
        pl.col("enllac_publicacio").alias("link_href"),
        pl.col("resultat").alias("status"),
        pl.col("termini_presentacio_ofertes").str.extract(r"\d{4}", 0).cast(pl.Int32).alias("year"),
        pl.col("valor_estimat_contracte").fill_null(0).alias("estimatedAmount"),
        pl.col("import_adjudicacio_sense_iva").alias("payableAmount"),
        pl.col("lloc_execucio").alias("city"),
        pl.col("durada_contracte").alias("duration"),
        pl.col("identificacio_adjudicatari").alias("cif_provider"),
        pl.col("denominacio_adjudicatari").alias("name_provider")
    )
len(raw)
duplicated = raw.unique()
len(duplicated)

```

```python
mapa_numeros = {
    "Adjudicació": "8",
    "Formalització": "9",
    "Desert": "3",
    "Desisitment": "4",
    "Renúncia":"5"
}

for palabra, digito in mapa_numeros.items():
    raw = raw.with_columns(
        pl.col("status").str.replace_all(palabra, digito)
    )


```

```python
mapa_numeros = {
    r"(?i)\buna?\b": "1",
    r"(?i)\bdos\b": "2",
    r"(?i)\btres\b": "3",
    r"(?i)\bquatre\b": "4",
    r"(?i)\bcinc\b": "5",
    r"(?i)\bsis\b": "6",
    r"(?i)\bset\b": "7",
    r"(?i)\bvuit\b": "8",
    r"(?i)\bnou\b": "9",
    r"(?i)\bdeu\b": "10"
}

pattern = r"(?i)(?:(\d+)\s*(?:HPO|VPP|habitatges?|allotjaments?|unifamiliars?|v\.)|(?:alojamientos?|viviendas?|unifamiliares?|v\.)\s*(\d+))"

v = raw.with_columns(
    pl.col("title").alias("title_clean")
)
for palabra, digito in mapa_numeros.items():
    v = v.with_columns(
        pl.col("title_clean").str.replace_all(palabra, digito)
    )

v = v.with_columns(
    pl.col("title_clean")
    .str.extract(pattern, 1)
    .fill_null(pl.col("title_clean").str.extract(pattern, 2))
    .cast(pl.Int32, strict=False)
    .alias("n_viviendas")
).drop("title_clean")
v
```

```python
f = v.with_columns(
    (
        pl.col("duration").str.extract(r"(\d+)", 1).cast(pl.Int64) * 365 +
        pl.col("duration").str.extract(r"\d+\D+(\d+)", 1).cast(pl.Int64) * 30 +
        pl.col("duration").str.extract(r"\d+\D+\d+\D+(\d+)", 1).cast(pl.Int64)
    ),
        pl.col("status").cast(pl.Int64),
        pl.col("payableAmount").str.split("||").list.get(0).cast(pl.Float64)
)
f
```

```python
city = raw.group_by("city").agg([
    pl.len().alias("count")
]).sort(by="count", descending=True)
city
```
